# Figures

```python
# Fig. 1: plots the average number of Medium or Hard encounters per adventuring day for the DMG
fig = go.Figure()

dmg = rules[[r['name'] for r in rules].index('DMG 2014')]

fig.add_trace(
    go.Scatter(
        x=dmg['level'],
        y=[6 for i in range(20)],
        line_color='rgba(0,0,0,0)',
        line_width=0,
        showlegend=False,
        hoverinfo='none',
    ),
)

fig.add_trace(go.Scatter(
    x=dmg['level'],
    y=[8 for i in range(20)],
    mode='none',
    fill='tonexty', 
    showlegend=False,
    fillcolor='rgba(100,100,100,0.3)',
    hoverinfo='none'
))

fig.add_trace(
    go.Scatter(
        x=dmg['level'],
        y=[dmg['Daily'][i]/(((dmg['Medium'][i] + dmg['Hard'][i]- 1) + (dmg['Hard'][i] + dmg['Deadly'][i] - 1))/4) for i in range(20)],
        name=dmg['name'],
        showlegend=False,
        line_color=COLOR_LIST[0],
        hovertemplate='<b>'+dmg['name']+'</b><br>level %{x:.0f}<br>encounters %{y:.1f}<extra></extra>',
    ),
)

print(np.mean([dmg['Daily'][i]/(((dmg['Medium'][i] + dmg['Hard'][i]- 1) + (dmg['Hard'][i] + dmg['Deadly'][i] - 1))/4) for i in range(20)]))


fig.update_layout(
    template=tfb.FIG_TEMPLATE,
    xaxis=dict(title_text='level', dtick=5, minor_dtick=1, range=[0,21]),
    yaxis=dict(title_text='encounters', range=[0,9], dtick=1),
    width=550, 
    height=450,
)

fig.show(config=tfb.FIG_CONFIG)

# save figures
if SAVEFIGS:
    fig.update_layout(autosize=True, width=None, height=None)
    tfb.save_fig_html(fig, format='large', name=f'./fig-dmg-daily-medium-to-hard-large')
    tfb.save_fig_html(fig, format='small', name=f'./fig-dmg-daily-medium-to-hard-small')
```

```python
# Fig. 2: shows the number of equivalent Easy encounters per adventuring day for each playtest and difficulty
fig = go.Figure()

for d, n in [('Easy',6.5),('Average',4),('Tough',2)]:
    x = [r['name'][3:] for r in rules if r['name'].startswith('PT')]
    y = [np.mean(n*np.array(r[d])/np.array(r['Easy'])) for r in rules if r['name'].startswith('PT')]
    fig.add_bar(
        x=x, 
        y=y,
        name=d,
        hovertemplate='<b>'+d+'</b><br>D&D Next %{x}<br>encounters %{y:.1f}<extra></extra>',
    )

fig.update_layout(
    template=tfb.FIG_TEMPLATE,
    xaxis=dict(title_text='D&D Next playtest'),
    yaxis=dict(title_text='Easy encounters', range=[0,7], dtick=1),
    width=650, 
    height=450,
)

fig.show(config=tfb.FIG_CONFIG)

# save figures
if SAVEFIGS:
    fig.update_layout(autosize=True, width=None, height=None)
    tfb.save_fig_html(fig, format='large', name=f'./fig-next-daily-encounters-large')
    tfb.save_fig_html(fig, format='small', name=f'./fig-next-daily-encounters-small')
```
